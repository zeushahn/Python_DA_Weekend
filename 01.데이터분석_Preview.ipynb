#### DataFrame    
- Pandas에서 사용하는 기본 데이터 타입
- 데이터프레임을 정의할때는 2차원 리스트를 전달 받는다.    


------
#### Pandas를 이용하여 DataFrame 만들기

```python
import pandas as pd
```

```python
# 행방향으로 DataFrame 만들기 : List
# List로 Data 정의
list01 = [
	[10, 20, 30],
	[40, 50, 60],
	[70, 80, 90]
]

# DataFrame 변환
df = pd.DataFrame(list01)
df
```

```python
# List Type으로 Dataset을 정의하여 DataFrame 생성

df = \
	pd.DataFrame(
		[
			[10, 20, 30],
			[40, 50, 60],
			[70, 80, 90]
		]
	)
df
```

> DataFrame은 index(rows)와 columns로 구성되어 있음.

```python
# Column이름이 Pandas의 기본 이름이므로 Column 이름 변경

df.columns = ['A', 'B', 'C']
df
```

#### DataFrame에서 값 추출하기

```python
# 열 기준으로 추출하기 : A번 열의 데이터 추출하기
df.loc[:,'A']
```

```python
# 열 기준으로 추출 : A,B번 열 데이터 추출
df.loc[:,['A','B']] # Column 이름으로 추출 
df.loc[:,'A':'B']  # Column의 범위로 추출 
df.loc[:,:'B'] # Column의 처음부터 시작하여 범위로 추출 
```

```python
# 범위를 정해 행 기준으로 추출하기 
df.loc[0,:]
```

```python
# 범위를 정해 행기준으로 추출하기
df.loc[:1,:]
df.loc[[0,1],:]
```

```python
# 행과 열의 이름으로 데이터 추출
print(df.loc[1,'B'])
```

```python
# DataFrame의 컬럼 이름과 index이름 출력하기
print(df.columns)
print(list(df.index))
```

#### Series
- Pandas에서 사용하는 기본 데이터 
- 1차원 리스트로 전달
- index는 자동 생성

```python
list01 = [10, 20, 30]
s1 = pd.Series(list01)
print(s1)
print(s1[1])
```

> Series는 List 대비 index이름이 정해지고 보여진다.

----
#### Dictionary({K:V})로 DataFrame 만들기

```python
dic01 = {
	"weight" : [ 80, 70.4, 65.5, 45.9, 51.2],
	"height" : [170,  180,  155,  143, 154],
	'type'   : ['f',  'n',  'n',  't', 't']
}

tbl = pd.DataFrame(dic01)
tbl
```

> dictionay의 key가 dataframe의 컬럼 이름으로 생성된다. 

```python
# weight열만 추출하기
tbl.loc[:,'weight']
```
