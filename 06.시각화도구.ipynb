#### Matplotlib
: 파이썬의 표준 시각화 도구

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("../Data/시도별 전출입 인구수.xlsx")
df.head()
```

```python
df.tail()
```

```python
df.shape
```

```python
df.info()
```

```python
# 행번호 0을 삭제
df.drop(
    0,
    axis='rows',
    inplace=True
)
df.head()
```

```python
# Index 재정리
df.reset_index(
    inplace=True,
    drop=True # 컬럼으로 생성하지 않고 삭제 한다.
)
df.head()
```

```python
# NaN 정리 : 현재 NaN을 상위의 값으로 채움

df.ffill(inplace=True) # bfill() : 하위값으로 채우기
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
# 컬럼 Type 정보
df.columns
```

#### 년도 컬럼의 데이터를 문자에서 숫자로 변환

```python
# 1970년의 데이터 내용 보기
df['1970'].unique()
```

```python
# 1970년의 전체 갯수
print(f"전체 갯수 : {len(df['1970'])}")

# 1970년의 -인 갯수
print(f"- 인 갯수 : {df[df['1970'] == '-'].shape[0]}")
```

```python
# - -> 0으로 변경
df.replace(
    '-',
    0,
    inplace=True
)
df.head()
```

```python
df.info()
```

---
#### 전출입 시각화

```python
# 서울에서 타지역으로 이동한 데이터만 추출
df_seoul = \
	df[(df.전출지별 == '서울특별시') & (df.전입지별 != '서울특별시')]
df_seoul.head()
```

```python
# 전출지별 컬럼 삭제
df_seoul.drop(
    '전출지별',
    axis='columns',
    inplace=True
)
df_seoul.head()
```

```python
# 전입지별 -> 전입지 변경
df_seoul.rename(
    {
        '전입지별' : '전입지'
	},
    axis='columns',
    inplace=True
)
df_seoul.head()
```

```python
# 전입지를 index로 변경
df_seoul.set_index(
    '전입지',
    inplace=True
)
df_seoul.head()
```

```python
# 서울에서 경기도로 이동한 인구의 선그래프로 시각화
sr_one = df_seoul.loc['경기도',:]
sr_one.head()
```

```python
# 선그래프
plt.plot(
    sr_one
)

plt.show()
```

### Ex
: 부산광역시에서 서울특별시로 전입한 그래프

```python
df_busan = \
	df[(df.전출지별 == "부산광역시") & (df.전입지별 == '서울특별시')]
df_busan
```

```python
# 전출지별 컬럼 삭제
df_busan.drop(
    '전출지별',
    axis='columns',
    inplace=True
)
df_busan
```

```python
# 전입지별 -> 전입지 수정 
df_busan.rename(
    {'전입지별' : '전입지'},
    axis='columns',
    inplace=True
)
df_busan
```

```python
# 전입지 컬럼을 index로 변경
df_busan.set_index(
    '전입지',
    inplace=True
)
df_busan
```

```python
df_busan.loc['서울특별시',:]
```

```python
# 그래프 사이즈 조정(가로 :14인치, 세로:5인치)
plt.figure(
    figsize=(14,5)
)

# 선그래프 
plt.plot(
    df_busan.loc['서울특별시',:]
)

# 차트 제목
plt.title("부산 -> 서울 이동")

# 축 이름
plt.xlabel("기간")
plt.ylabel("이동 인구수")

# x축 눈금 라벨 회전하기
plt.xticks(
    rotation = 45
)

# y축 눈금 라벨 회전하기
plt.yticks(
    rotation=-15
)

# 범례
plt.legend(
    labels = ['부산 -> 서울'],
    loc='best' # best, 'upper left' , 'lower right'
)

plt.show()
```

#### 그래프에 스타일 적용하기

```python
# Style 서식 지정
plt.style.use('ggplot')

# 그래프 사이즈
plt.figure(
    figsize=(14, 5)
)

# 선그래프
plt.plot(
    sr_one,
    marker = 'o', # +, ^, *
    markersize=10,
    zorder=1 # scatter와 z축 우선순위 변경 
)

# 차트 제목
plt.title(
    "서울 -> 경기 이동\n",
    size=20
)

# 축이름
plt.xlabel(
    "기간",
    size=12
)
plt.ylabel(
    "이동 인구수",
    size=12
)

# x축 눈금 회전하기
plt.xticks(
    rotation=45
)

# y축 눈금 회전하기
plt.yticks(
    rotation=45
)

# 범례
plt.legend(
    labels = ['서울->경기'],
    loc='best',
    fontsize=20
)

# -------------------------
# y축 범위 설정
plt.ylim(0,700000)

# 1995년 데이터 표시
plt.annotate(
    f'{sr_one['1995']:,}명',
    xy=(1995-1970, sr_one['1995']+20000),
    va='baseline',
    ha='center'
)

# 2017년 데이터 표시
plt.annotate(
    f'{sr_one['2017']:,}명',
    xy=(2017-1970, sr_one['2017']-40000),
    va='baseline',
    ha='center'
)

# 화살표 annotate
plt.annotate(
    f'{sr_one['1970']:,}명',
    xy=(1970-1970, sr_one['1970']), # 화살표의 머리부분
    xytext=(1970-1970, sr_one['1970']+100000), # 화살표의 꼬리부분
    xycoords='data',
    arrowprops=dict(arrowstyle='->', color='blue', lw=3),
    va='baseline',
    ha='center'
)

# 특정 위치의 marker만 틀리게 하기
plt.scatter(
    1995-1970,
    sr_one['1995'],
    color = 'blue',
    s=500,
    marker='o'
)


plt.show()
```
