----
#### Boolean Indexing

```python
# 타이타닉호에 승선한 남성 인원수는?
print(len(train[train.Sex == 'male']))
print(train[train.Sex == 'male'].shape[0])
print(train[train.Sex == 'male'].Survived.count()) # 결측치는 계산이 않됨 
```

```python
# 운임(Fare)이 20달러 이상을 지불한 사람의 수?
train[train.Fare >= 20].shape[0]
```

```python
# 운임(Fare)이 평균 운임 이상으로 지급한 사람의 수?
train[train.Fare >= train.Fare.mean()].shape[0]
```

```python
# 운임(Fare)이 평균 운임이상으로 지급한 사람들 중 1등급 객실을 사용한 인원수?
# &(and), |(or), !(not)
train[(train.Fare >= train.Fare.mean()) & (train.Pclass == 1)].shape[0]
```

```python
# 최대 운임을 지불한 사람들은?
train[train.Fare == train.Fare.max()]
```

```python
# 항구 (Embarked)
# C항구에서 탑승한 인원수
print(f"C항구 탑승인원수 : {train[train.Embarked == 'C'].shape[0]}")
# S항구에서 탑승한 인원수
print(f"S항구 탑승인원수 : {train[train.Embarked == 'S'].shape[0]}")
# Q항구에서 탑승한 인원수
print(f"Q항구 탑승인원수 : {train[train.Embarked == 'Q'].shape[0]}")
```

```python
for emb in ['C', 'S', 'Q']:
    print(f"{emb}항구 탑승인원수 : {train[train.Embarked == emb].shape[0]}")
```

```python
for emb in train.Embarked.unique():
    print(f"{emb}항구 탑승인원수 : {train[train.Embarked == emb].shape[0]}")
```

```python
# 항구별 탑승 인원수 (NaN도 포함)
for emb, count in train.Embarked.value_counts(dropna=False).items():
    port = "항구 미상(NaN)" if pd.isna(emb) else f"{emb}항구"
    print(f"{port} 탑승인원수 : {count}")
```
